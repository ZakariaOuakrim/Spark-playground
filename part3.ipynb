```python
from pyspark.sql import SparkSession

spark=SparkSession.builder.appName("Practice").getOrCreate()
```

```python
df=spark.read.csv("test2.csv",header=True,inferSchema=True)
df.show()
```

## filter operation

```python
# filter salary of people who have salary>=6000
df.filter("salary>=6000").show()
```

```python
df.filter("salary>=6000").select("name","age").show()
```

```python
df.filter(df["Salary"]>=6000).show()
```

```python
df.filter((df["Salary"]>=6000) & (df["age"]>21)).show()
```
