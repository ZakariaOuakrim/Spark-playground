```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Practice").getOrCreate()
```

```python
df=spark.read.csv("test2.csv",header=True,inferSchema=True)
df.show()
```

```python
##drop columns 
df.drop("Name").show()
```

```python
df.show()
```

```python
# wherever there is a null value it's getting deleted 
df.na.drop().show()
```

```python
# if all of the line has Null
df.na.drop(how="all").show()
```

```python
# if less than 3 non null values in a row then remove it  
df.na.drop(how="any",thresh=3).show()
```

```python
# subset = drop Null only in one columns 
df.na.drop(how="any",subset="salary").show()
```

```python
#filling missing value you can add this part ,["experience","age"] to add missing values only on those two columns
df.na.fill("missing values",["experience","age"]).show()
```

```python
df.show()
df.na.fill(12,["age"]).show()
```
